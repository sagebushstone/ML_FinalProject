```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.tree import DecisionTreeClassifier
import numpy as np
```

```python
spotify_df = pd.read_csv("SpotifyDataset.csv")
spotify_reduced_df = spotify_df[["energy", "duration_ms", "liveness", "valence", "mode", "explicit", "speechiness", "acousticness", "instrumentalness", "liveness", "popularity"]]
spotify_reduced_df["explicit"] = spotify_reduced_df["explicit"].map({True: 1, False: 0})
x = spotify_reduced_df
y = spotify_df[['track_genre']]
x.head(5)
```

```python
x_train,x_test,y_train,y_test =  train_test_split(x,y,test_size=0.2,random_state=0)

from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
model = DecisionTreeClassifier()
model.fit(x_train, y_train)
predictions = model.predict(x_test)
accuracy = metrics.accuracy_score(y_test, predictions)
```

```python
accuracy
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(x_train, np.ravel(y_train))
predictions = model.predict(x_test)
accuracy = metrics.accuracy_score(y_test, predictions)
```
